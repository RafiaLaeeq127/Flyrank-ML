```python
from datasets import load_dataset

# Specify 'fact_content_daily_performance' subset configuration
dataset = load_dataset("FlyRank/internship-warehouse", "fact_content_daily_performance")
train = dataset['train']

# Fast batched filter (uses Arrow under the hood, not row-by-row Python loop)
def month_mask(batch):
    return [d.strftime('%Y-%m') == '2025-01' for d in batch['report_date']]

month_ds = train.filter(month_mask, batched=True)
month_df = month_ds.to_pandas()

print(f"Pulled {len(month_df)} rows for 2025-01")
```

```python
# --- Query A: Grain check ---
grain_check = month_df.groupby(['client_hash_id','content_hash_id','report_date']).size()
print(f"Max rows per client-content-day combo: {grain_check.max()} (should be 1 if grain is correct)")

# --- Query B: Slice row count and date span ---
print(f"Total rows in slice: {len(month_df)}")
print(f"Date span: {month_df['report_date'].min()} to {month_df['report_date'].max()}")

# --- Query C: Availability filter ---
available = month_df[month_df['gsc_data_available'] == True]
print(f"Rows surviving gsc_data_available IS TRUE: {len(available)} out of {len(month_df)} "
      f"({len(available)/len(month_df)*100:.1f}%)")
```

```python
# Keep only rows with real impressions (needed for CTR to make sense)
usable = available[available['gsc_impressions'] > 0].copy()
usable['ctr'] = usable['gsc_clicks'] / usable['gsc_impressions']
print(f"Usable rows (impressions > 0): {len(usable)}")

features = usable[['client_hash_id','content_hash_id','report_date']].copy()

features['gsc_avg_position'] = usable['gsc_avg_position']
# knowable at the decision moment because: position is logged directly by Search Console at serve time

features['gsc_impressions'] = usable['gsc_impressions']
# knowable at the decision moment because: it's a running count already recorded for that day

features['ctr_rolling_3day'] = usable.sort_values('report_date').groupby('content_hash_id')['ctr'].transform(
    lambda x: x.rolling(3, min_periods=1).mean()
)
# knowable at the decision moment because: only uses the past 3 days, no future data involved

features['sessions_organic'] = usable['sessions_organic']
# knowable at the decision moment because: already-recorded traffic for that day

features['client_has_ga4'] = usable['client_has_ga4']
# knowable at the decision moment because: static client property, known regardless of outcome

features.head(10)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

target = (usable['ctr'] > usable['ctr'].quantile(0.8)).astype(int)

# --- WITH the trap: gsc_clicks directly determines the label ---
X_leaky = features[['gsc_avg_position','gsc_impressions','ctr_rolling_3day']].copy()
X_leaky['gsc_clicks'] = usable['gsc_clicks']  # <-- THE TRAP

model_leaky = LogisticRegression(max_iter=1000).fit(X_leaky.fillna(0), target)
score_leaky = roc_auc_score(target, model_leaky.predict_proba(X_leaky.fillna(0))[:,1])
print(f"WITH leakage column: AUC = {score_leaky:.3f}  <- suspiciously close to 1.0")

# --- Remove the trap, keep the honest number ---
X_honest = features[['gsc_avg_position','gsc_impressions','ctr_rolling_3day']].copy()
model_honest = LogisticRegression(max_iter=1000).fit(X_honest.fillna(0), target)
score_honest = roc_auc_score(target, model_honest.predict_proba(X_honest.fillna(0))[:,1])
print(f"WITHOUT leakage column: AUC = {score_honest:.3f}  <- the honest number")

print(f"""
Lesson: adding gsc_clicks (which directly determines the CTR-based label)
pushed AUC from {score_honest:.3f} to {score_leaky:.3f} — a textbook leakage
trap. Dropping it keeps the model honest about what it can predict ahead of time.
""")
```

## Summary

- Grain confirmed: 1 row = 1 client-content-day (verified on 2025-01, 1297 rows)
- Data span available in this slice: 2025-01-27 to 2025-01-31
- 100% of rows had gsc_data_available = True
- Leakage test: including gsc_clicks (which the CTR label is derived from)
  pushed AUC from 0.864 (honest) to 1.000 (leaky) — confirms the model learns
  real signal from position/impressions/rolling-CTR, not just memorizing the label.
