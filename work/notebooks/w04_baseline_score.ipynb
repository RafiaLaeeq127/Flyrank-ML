```python
import pandas as pd
import numpy as np
import os

# Correct path identified from searching the repository
data_path = 'data/raw/content_refresh_anonymized.csv'

df = pd.read_csv(data_path)
print("Columns in dataset:", df.columns.tolist())
df.head()
```

```python
# Print all column names in full
for i, col in enumerate(df.columns):
    print(f"{i}: {col}")
```

```python
# Signal 1: Staleness (days_since_last_update) vs Trend / Performance
# Bin staleness into standard operational tiers
staleness_bins = [-1, 30, 90, 180, 365, float("inf")]
staleness_labels = ["0-30d (Fresh)", "31-90d (Recent)", "91-180d (Aging)", "181-365d (Stale)", "365d+ (Legacy)"]

df["staleness_bucket"] = pd.cut(df["days_since_last_update"], bins=staleness_bins, labels=staleness_labels)

# Group by bucket and compute sample count n, average impressions, and average sessions
signal_1_table = df.groupby("staleness_bucket", observed=False).agg(
    n=("content_id", "count"),
    avg_sessions_last_30d=("sessions_last_30d", "mean"),
    avg_sessions_prev_30d=("sessions_prev_30d", "mean"),
    avg_impressions_last_30d=("impressions_last_30d", "mean"),
    avg_trend_pct=("trend_pct", "mean")
).reset_index()

print("=== Signal 1: Staleness Bucket Table ===")
print(signal_1_table)
```

```python
# Signal 2: Position vs CTR (FlyRank CTR-fix heuristic)
pos_bins = [0, 3, 10, 20, 50, 100]
pos_labels = ["Top 3 (1-3)", "Strike Zone (4-10)", "Page 2 (11-20)", "Page 3-5 (21-50)", "Deep (51+)"]

df["pos_bin"] = pd.cut(df["avg_position"], bins=pos_bins, labels=pos_labels)

signal_2_table = df.groupby("pos_bin", observed=False).agg(
    n=("content_id", "count"),
    mean_ctr=("ctr", "mean"),
    median_ctr=("ctr", "median"),
    avg_impressions_90d=("impressions_90d", "mean")
).reset_index()

print("=== Signal 2: CTR vs Position Bucket Table ===")
print(signal_2_table)
```

```python
import os

# Ensure the output directory exists
os.makedirs("work/outputs", exist_ok=True)

# Define the baseline heuristic rule: Stale High-Impression Content Refresh
# Logic: Content not updated in >90 days, but still driving search impressions
def apply_baseline_rule(row):
    # Condition: Aging or stale content with meaningful impression volume
    if row["days_since_last_update"] > 90 and row["impressions_90d"] > 50:
        # Score formula: impressions weighted by staleness (normalized days)
        score = float(row["impressions_90d"] * (row["days_since_last_update"] / 100.0))
        return pd.Series([score, "STALE_HIGH_IMPRESSIONS", "REFRESH_AND_UPDATE_CONTENT"])

    # Strike-zone low CTR candidate
    elif (4.0 <= row["avg_position"] <= 10.0) and (row["impressions_90d"] > 100) and (row["ctr"] < 0.05):
        score = float(row["impressions_90d"] * (11.0 - row["avg_position"]))
        return pd.Series([score, "STRIKE_ZONE_LOW_CTR", "OPTIMIZE_TITLE_AND_SNIPPET"])

    return pd.Series([0.0, "NO_ACTION", "NONE"])

# Apply rule
df[["score", "reason_code", "action_label"]] = df.apply(apply_baseline_rule, axis=1)

# Filter for actionable items and rank descending by score
ranked_queue = df[df["score"] > 0].sort_values(by="score", ascending=False).reset_index(drop=True)

# Select relevant columns for the final queue
output_cols = [
    "content_id", "client_id", "score", "reason_code", "action_label",
    "days_since_last_update", "impressions_90d", "clicks_90d", "ctr", "avg_position"
]
final_output = ranked_queue[output_cols]

# Save to work/outputs/baseline_action_score.csv
csv_path = "work/outputs/baseline_action_score.csv"
final_output.to_csv(csv_path, index=False)

print(f"Successfully generated {len(final_output)} prioritized actions!")
print(f"Saved to: {csv_path}")
```
