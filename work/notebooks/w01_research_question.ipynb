## Week 1 Research Question — CTR / Engagement Opportunity Scoring

**Search question:** Which pages have high impressions but low click-through rate
relative to their ranking position — and are therefore the best candidates for
title/meta-description optimization?

**Unit of analysis:** Page

**Output:** A ranked list of pages with an "opportunity score" — high impressions,
low CTR relative to their position tier.

**Action someone could take:** An SEO/content team member prioritizes rewriting
titles and meta descriptions for the top-scored pages first, instead of guessing
which pages need attention.

**Cost of a wrong recommendation:** If the model flags the wrong pages, the team
wastes time optimizing content that was never underperforming, while pages that
genuinely need attention keep leaking traffic — a real cost in effort and missed clicks.

**Why data/ML helps:** With thousands of pages, no one can manually check CTR-vs-position
patterns for each one. A simple score that adjusts for the fact that top positions
naturally get higher CTR (so raw CTR comparison alone is misleading) surfaces the
real outliers automatically.


```python
# Load the same starter dataset used in notebooks 01/02
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")  # adjust path/filename to match yours

# Check actual column names if unsure
print(df.columns.tolist())

# Number 1: how many pages have high impressions but low CTR
median_ctr = df['ctr'].median()
high_impressions = df['impressions_90d'] > df['impressions_90d'].median()
low_ctr = df['ctr'] < median_ctr
opportunity_pages = df[high_impressions & low_ctr]

print(f"Pages with high impressions + below-median CTR: {len(opportunity_pages)} "
      f"out of {len(df)} total pages ({len(opportunity_pages)/len(df)*100:.1f}%)")

# Number 2: correlation between avg_position and CTR
corr = df['avg_position'].corr(df['ctr'])
print(f"Correlation between avg_position and CTR: {corr:.3f}")

# Number 3: CTR spread within a similar position tier (e.g. position 5-10)
mid_tier = df[(df['avg_position'] >= 5) & (df['avg_position'] <= 10)]
print(f"CTR range for pages ranked 5-10: min={mid_tier['ctr'].min():.3f}, "
      f"max={mid_tier['ctr'].max():.3f}, std={mid_tier['ctr'].std():.3f}")
```

```python
#interpretation
Pages with high impressions + below-median CTR: 3309 out of 30000 total pages (11.0%)
Correlation between avg_position and CTR: -0.073
CTR range for pages ranked 5-10: min=0.000, max=100.000, std=2.595
```
